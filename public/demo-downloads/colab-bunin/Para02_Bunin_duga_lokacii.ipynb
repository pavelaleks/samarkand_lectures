## Шаг 1. Дуга тональности
Текст режется на N равных отрезков. В каждом считаем слова из списков POS и NEG (по леммам) на 1000 слов.
**Это не «чувство автора»**, а распределение оценочной лексики. Ваш словарь = ваша гипотеза.

```python
#@title 📈 Словари и дуга { display-mode: "form" }
POS = "счастье, счастливый, любовь, любить, нежный, радость, радостный, светлый, милый, весёлый, смеяться, улыбаться, тепло, прелестный"  #@param {type:"string"}
NEG = "смерть, умереть, убить, война, горе, страшный, боль, тоска, слёзы, плакать, холод, холодный, разлука, одиночество, гроб, могила"  #@param {type:"string"}
N_SEGMENTS = 8  #@param {type:"slider", min:4, max:20, step:1}
for title, text in texts.items():
    arc_df = arc(text, POS, NEG, n=N_SEGMENTS, title=title)
arc_df[["отрезок", "POS", "NEG", "баланс /1000", "POS-слова", "NEG-слова"]]
```

**Как читать:** найдите отрезок с минимумом и максимумом. Посмотрите колонки «POS-слова/NEG-слова» — **не обманул ли словарь?**
Например, «холодный» в названии сезона — это ещё не горе. Затем прочитайте этот отрезок целиком:

```python
#@title 📖 Прочитать отрезок целиком { display-mode: "form" }
SEGMENT_NO = 4  #@param {type:"integer"}
TEXT_NO = 1  #@param {type:"integer"}
_t = list(texts.values())[TEXT_NO - 1]
print(segments(_t, n=N_SEGMENTS)[SEGMENT_NO - 1])
```
