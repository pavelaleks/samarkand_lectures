```python
import re, requests, matplotlib.pyplot as plt
import pandas as pd

BASE = "https://pavelaleks.github.io/samarkand_lectures/demo-corpora/bunin-temnye-allei/txt/"
ho = requests.get(BASE + "holodnaya-osen.txt", timeout=60).text
cp = requests.get(BASE + "chistyj-ponedelnik.txt", timeout=60).text
print(len(ho.split()), len(cp.split()))
```

## Дуга по чанкам (эвристика POS/NEG)


```python
POS = {"счаст", "люб", "светл", "нежн", "рад", "мирн", "тепл"}
NEG = {"смерть", "убий", "холод", "страш", "боль", "тоск", "горе", "войн", "разлук"}

def tokenize(t):
    return re.findall(r"[А-Яа-яЁё]+", t.lower())

def chunk_text(text, n=8):
    words = text.split()
    size = max(len(words) // n, 1)
    return [" ".join(words[i:i+size]) for i in range(0, len(words), size)][:n]

def score(chunk):
    toks = tokenize(chunk)
    p = sum(1 for w in toks if any(w.startswith(s) for s in POS))
    n = sum(1 for w in toks if any(w.startswith(s) for s in NEG))
    return p - n

chunks = chunk_text(ho, 8)
scores = [score(c) for c in chunks]
plt.figure(figsize=(8, 3))
plt.plot(range(1, len(scores)+1), scores, marker="o")
plt.axhline(0, color="gray", lw=1)
plt.title("Учебная дуга · Холодная осень"); plt.xlabel("чанк"); plt.ylabel("POS−NEG")
plt.tight_layout(); plt.show()
print(list(enumerate(scores, 1)))
```

## Учебная карта мест


```python
PLACES_HO = ["деревня", "сад", "дом", "станция", "война"]
PLACES_CP = ["Москва", "Новодевичий", "Замоскворечье", "Ордынка", "церковь", "ресторан"]

def place_counts(text, places):
    low = text.lower()
    return {p: low.count(p.lower()) for p in places}

print("Холодная осень", place_counts(ho, PLACES_HO))
print("Чистый понедельник", place_counts(cp, PLACES_CP))
```
