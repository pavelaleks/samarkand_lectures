## Режим: thesis или demo


```python
import re
from pathlib import Path

MODE = "demo"  # смените на "thesis" после загрузки своих .txt

# --- thesis: имена загруженных файлов ---
THESIS_FILES = ["glava1.txt", "glava2.txt"]  # поправьте

# --- demo: Чехов ---
import requests
CHEKHOV = "https://pavelaleks.github.io/samarkand_lectures/demo-corpora/chekhov-rasskazy/txt/"
DEMO = {
    "Дама с собачкой": "dama-s-sobachkoj.txt",
    "Ионыч": "ionyich.txt",
    "Студент": "student.txt",
}

texts = {}
if MODE == "thesis":
    for fname in THESIS_FILES:
        p = Path(fname)
        assert p.exists(), f"Нет файла {fname} — загрузите слева в Colab"
        texts[p.stem] = p.read_text(encoding="utf-8")
else:
    for title, fname in DEMO.items():
        r = requests.get(CHEKHOV + fname, timeout=60)
        r.raise_for_status(); r.encoding = "utf-8"
        texts[title] = r.text

for k, v in texts.items():
    print(k, len(v), "символов")
```

## Паспорт корпуса (заполните ячейку-Markdown копией таблицы)


```python
passport = []
for title, text in texts.items():
    passport.append({
        "единица": title,
        "символов": len(text),
        "слов_грубо": len(text.split()),
        "издание_или_источник": "УКАЖИТЕ",
        "зачем_в_диссертации": "УКАЖИТЕ",
        "что_исключили": "УКАЖИТЕ",
    })
import pandas as pd
pd.DataFrame(passport)
```

## Учебная очистка


```python
def clean_text(raw):
    t = raw
    t = t.replace("\xa0", " ")
    t = re.sub(r"\[\d+\]", " ", t)          # сноски [1]
    t = re.sub(r"(?m)^\s*\d+\s*$", " ", t) # номера страниц
    t = re.sub(r"https?://\S+", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

cleaned = {k: clean_text(v) for k, v in texts.items()}
for k in cleaned:
    print(k, "было", len(texts[k]), "стало", len(cleaned[k]))

# сохраните один clean-файл
sample_key = next(iter(cleaned))
Path(f"{sample_key}_clean.txt").write_text(cleaned[sample_key], encoding="utf-8")
print("saved", f"{sample_key}_clean.txt")
```
