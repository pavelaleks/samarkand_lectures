## Свой словарь
Соберите **10–20 слов в каждый список под вашу тему**: для сатиры — «смешной/глупый», для элегии — «печаль/утрата», для идиллии — «тихий/покой».
Хороший приём: выпишите слова из 2–3 ключевых сцен вашего текста.

```python
#@title 📈 Словари и дуга { display-mode: "form" }
POS = "счастье, счастливый, любовь, любить, нежный, радость, весёлый, молодой, свежий, прекрасный, милый, смеяться, улыбаться"  #@param {type:"string"}
NEG = "скучный, скука, тоска, страшный, грустный, печальный, плакать, слёзы, старый, пошлый, унизительный, ложь, тяжёлый, обман"  #@param {type:"string"}
N_SEGMENTS = 10  #@param {type:"slider", min:4, max:30, step:1}
title0 = list(texts)[0]
arc_df = arc(texts[title0], POS, NEG, n=N_SEGMENTS, title=title0)
lo, hi = arc_df["баланс /1000"].idxmin(), arc_df["баланс /1000"].idxmax()
print(f"Минимум — отрезок {lo + 1}, максимум — отрезок {hi + 1}")
arc_df.loc[[lo, hi], ["отрезок", "баланс /1000", "POS-слова", "NEG-слова", "начало отрезка"]]
```

```python
#@title 📖 Прочитать отрезок-минимум целиком { display-mode: "form" }
print(segments(texts[title0], n=N_SEGMENTS)[lo])
```

```python
#@title 🔎 Проверка цитатой: слово в контексте (конкорданс) { display-mode: "form" }
#@markdown Слова через запятую. Звёздочка = любое продолжение: `памят*`. Считается по леммам: `вспомнить` найдёт «вспомнила», «вспомню».
WORDS = "тоска, скучный, скука"  #@param {type:"string"}
kwic(texts, WORDS, width=70, limit=25)
```
