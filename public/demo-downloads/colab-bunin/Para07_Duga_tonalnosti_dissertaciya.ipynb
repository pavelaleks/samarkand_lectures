### Свой текст диссертации
В Colab: значок папки слева → загрузите `.txt` (UTF-8) фрагментов/глав, которые идут в магистерскую.
Дальше в коде поставьте `MODE = "thesis"` и перечислите имена файлов.


```python
import re, requests
from pathlib import Path
import matplotlib.pyplot as plt

MODE = "demo"
THESIS_FILE = "glava1.txt"
url = "https://pavelaleks.github.io/samarkand_lectures/demo-corpora/chekhov-rasskazy/txt/skuchnaya-istoriya.txt"

if MODE == "thesis":
    text = Path(THESIS_FILE).read_text(encoding="utf-8")
else:
    r = requests.get(url, timeout=60); r.encoding = "utf-8"; text = r.text

# ЗАМЕНИТЕ словари под вашу тему (ирония, сатира, траур, идиллия…)
POS = {"счаст", "люб", "радост", "светл", "нежн"}
NEG = {"смерть", "тоск", "страш", "боль", "горе", "скук", "плох"}

def tokenize(t):
    return re.findall(r"[А-Яа-яЁё]+", t.lower())

def chunk_text(text, n=10):
    words = text.split(); size = max(len(words)//n, 1)
    return [" ".join(words[i:i+size]) for i in range(0, len(words), size)][:n]

def score(chunk):
    toks = tokenize(chunk)
    p = sum(1 for w in toks if any(w.startswith(s) for s in POS))
    n = sum(1 for w in toks if any(w.startswith(s) for s in NEG))
    return p - n

chunks = chunk_text(text, 10)
scores = [score(c) for c in chunks]
plt.figure(figsize=(9, 3))
plt.plot(range(1, len(scores)+1), scores, marker="o")
plt.axhline(0, color="gray"); plt.title("Дуга · ваш текст"); plt.xlabel("чанк")
plt.tight_layout(); plt.show()

# найдите чанк-минимум и прочитайте глазами:
i_min = scores.index(min(scores))
print("мин чанк", i_min+1, "score", scores[i_min])
print(chunks[i_min][:500])
```
