```python
import re, requests
from collections import Counter
import pandas as pd

# DEMO: Чистый понедельник. Для диссертации — Path("...").read_text()
text = requests.get("https://pavelaleks.github.io/samarkand_lectures/demo-corpora/bunin-temnye-allei/txt/chistyj-ponedelnik.txt", timeout=60).text

# СВОИ списки под диссертацию:
PLACES = ["Москва", "Ордынка", "Замоскворечье", "Новодевичий", "Кремль", "церковь", "монастырь", "ресторан"]
PERSONS = ["он", "она"]  # лучше заменить на имена/прозвища вашего текста

low = text

def count_span(span):
    return len(re.findall(re.escape(span), text, flags=re.I))

rows = []
for p in PLACES:
    rows.append({"тип": "место", "якорь": p, "вхождений": count_span(p)})
for p in PERSONS:
    rows.append({"тип": "персона", "якорь": p, "вхождений": count_span(p)})
df = pd.DataFrame(rows).sort_values("вхождений", ascending=False)
df
```

```python
# простая «карта» без геокодера — бар по местам
m = df[df["тип"]=="место"].set_index("якорь")["вхождений"]
m.plot(kind="barh", figsize=(7, 4), title="Якоря мест (не география!)")
import matplotlib.pyplot as plt
plt.tight_layout(); plt.show()
```
