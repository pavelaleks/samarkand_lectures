## Правило ребра — это ваше методологическое решение
Узел = персонаж (или локус, мотив). Ребро = два узла **упомянуты в одном предложении (или абзаце)**.
Слева от двоеточия — имя узла, справа — все обозначения героя: имя, уменьшительные, «я» рассказчика.
В «Натали» рассказ ведётся от первого лица, поэтому рассказчик = «я, Мещерский, Виталий».

```python
#@title 🕸️ Узлы и правило ребра { display-mode: "form" }
NODES = "Я (Мещерский): я, Мещерский, Виталий | Соня: Соня | Натали: Натали, Наташа | Станкевич: Станкевич | Москва: Москва | Воронеж: Воронеж | сад: сад"  #@param {type:"string"}
WINDOW = "предложение"  #@param ["предложение", "абзац"]
MIN_WEIGHT = 2  #@param {type:"slider", min:1, max:10, step:1}
G, edges = cooccurrence(texts, NODES, window=WINDOW, min_weight=MIN_WEIGHT)
draw_graph(G)
centrality(G)
```

**Как читать:** *взвешенная степень* — насколько часто узел «вместе» с другими; *посредничество* — через кого проходят связи (герой-медиатор).
