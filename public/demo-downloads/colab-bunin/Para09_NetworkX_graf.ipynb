```python
!pip -q install networkx
import re, requests, itertools
from collections import Counter
import networkx as nx
import matplotlib.pyplot as plt

text = requests.get("https://pavelaleks.github.io/samarkand_lectures/demo-corpora/bunin-temnye-allei/txt/chistyj-ponedelnik.txt", timeout=60).text
# Для диссертации замените NODES на своих героев/локусы/мотивы
NODES = ["он", "она", "Москва", "церковь", "театр", "ресторан", "монастырь"]

sents = re.split(r"[.!?…]+", text)
weights = Counter()
for s in sents:
    present = [n for n in NODES if re.search(r"\b" + re.escape(n) + r"\b", s, flags=re.I)]
    for a, b in itertools.combinations(sorted(set(present)), 2):
        weights[(a, b)] += 1

G = nx.Graph()
for (a, b), w in weights.items():
    if w >= 2:  # порог — часть метода
        G.add_edge(a, b, weight=w)

plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx_nodes(G, pos, node_color="#99f6e4", node_size=1400)
nx.draw_networkx_labels(G, pos, font_size=9)
nx.draw_networkx_edges(G, pos, width=[G[u][v]["weight"] for u, v in G.edges()])
plt.axis("off"); plt.title("Совстречаемость в предложении (порог ≥2)")
plt.tight_layout(); plt.show()
print(sorted(G.edges(data=True), key=lambda e: -e[2]["weight"])[:10])
```

## Сдать: правило ребра + порог + одно **спорное** ребро с цитатой-предложением.
