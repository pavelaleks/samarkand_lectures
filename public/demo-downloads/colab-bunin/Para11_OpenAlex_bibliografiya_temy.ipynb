```python
import requests, time
from pathlib import Path

# === ЗАПОЛНИТЕ ПОД ДИССЕРТАЦИЮ ===
QUERIES = [
    '"Ваш Автор" metaphor',           # пример: точнее с кавычками
    'Ваш Автор AND memory',
    'Ваш_ключевой_термин AND literature',
]
# демо, если ещё не сформулировали:
QUERIES = [
    '"Ivan Bunin" emigration',
    '"Dark Avenues" OR "Тёмные аллеи"',
    'Bunin AND memory AND short story',
]

MAILBOX = "student@example.com"  # для вежливого пула OpenAlex
headers = {"User-Agent": f"SamGU-DH-Course/1.0 (mailto:{MAILBOX})"}

lines = ["OpenAlex raw for thesis bibliography", ""]
for q in QUERIES:
    r = requests.get(
        "https://api.openalex.org/works",
        params={"search": q, "per-page": 15, "sort": "relevance_score:desc"},
        headers=headers,
        timeout=60,
    )
    r.raise_for_status()
    data = r.json()
    lines.append(f"=== QUERY: {q} | total≈{data.get('meta', {}).get('count')} ===")
    for w in data.get("results", []):
        year = w.get("publication_year")
        title = w.get("display_name")
        doi = (w.get("doi") or "")
        lines.append(f"{year}\t{title}\t{doi}")
    lines.append("")
    time.sleep(0.3)

Path("openalex_thesis_raw.txt").write_text("\n".join(lines), encoding="utf-8")
print("saved openalex_thesis_raw.txt, lines:", len(lines))
print("\n".join(lines[:40]))
```
