### Как работать (3 правила)
1. **Файл → Сохранить копию на Диске.** Иначе ваши правки пропадут.
2. Запускайте ячейки **по очереди сверху вниз** кнопкой ▶ слева от ячейки. Код спрятан — менять нужно **только поля формы** (справа в ячейке).
3. Сначала запустите всё **как есть** — это демо на Бунине, результат заведомо осмысленный. Потом в ячейке «Тексты» выберите **«Мои файлы»** и пройдите те же шаги на материале своей диссертации.

> Что-то сломалось? **Среда выполнения → Перезапустить сеанс**, затем ячейки сверху вниз.
> Красная ошибка — скопируйте **последние 3 строки** преподавателю или в бесплатный чат ИИ с вопросом «что это значит для новичка?».

### Как составить запросы
- OpenAlex индексирует в основном **англоязычную** науку — давайте запросы и по-английски, и по-русски.
- Кавычки — точная фраза: `"Dark Avenues"`. Несколько запросов — через `|`.
- Лучше 3–5 узких запросов под **вашу** проблему, чем один «всё про автора».
- Без ключа лимит маленький (вся группа в одной сети может его исчерпать). **Бесплатный ключ:** зарегистрируйтесь на [openalex.org](https://openalex.org/settings/api) и вставьте в поле API_KEY.

```python
#@title 🔍 Запросы { display-mode: "form" }
QUERIES = '"Ivan Bunin" | "Dark Avenues" Bunin | Bunin memory love short story | Бунин "Темные аллеи"'  #@param {type:"string"}
API_KEY = ""  #@param {type:"string"}
PER_QUERY = 25  #@param {type:"slider", min:5, max:100, step:5}
FROM_YEAR = 0  #@param {type:"integer"}
bib = openalex(QUERIES, api_key=API_KEY, per_query=PER_QUERY, from_year=FROM_YEAR or None)
pd.set_option("display.max_colwidth", 90)
bib.head(20)
```

```python
#@title 📈 По годам и самые цитируемые { display-mode: "form" }
plot_years(bib, "Публикации по запросам")
bib.sort_values("цитирований", ascending=False).head(10)[["год", "название", "авторы", "издание", "цитирований", "doi"]] if not bib.empty else bib
```
