# Пара 5+. Бесплатные API книг: Gutendex + Open Library

Без ключей. Ищем издания/переводы вокруг Bunin и строим простой график по годам.


```python
import requests, pandas as pd, matplotlib.pyplot as plt
from collections import Counter

# 1) Gutendex (Project Gutenberg)
g = requests.get("https://gutendex.com/books", params={"search": "Bunin"}, timeout=60).json()
print("Gutendex count", g.get("count"))
for b in g.get("results", [])[:8]:
    print("-", b.get("id"), b.get("title"), "| langs:", b.get("languages"))

# 2) Open Library
ol = requests.get("https://openlibrary.org/search.json", params={"author": "Ivan Bunin", "limit": 50}, timeout=60).json()
print("\nOpen Library numFound", ol.get("numFound"))
years = []
for doc in ol.get("docs", []):
    y = doc.get("first_publish_year")
    if y and 1900 <= int(y) <= 2025:
        years.append(int(y))
    print("-", y, doc.get("title"))

cnt = Counter(years)
series = pd.Series(cnt).sort_index()
plt.figure(figsize=(9,4))
series.plot(kind="bar")
plt.title("Open Library: first_publish_year для выдачи author=Ivan Bunin")
plt.xlabel("год"); plt.ylabel("число записей в выборке")
plt.tight_layout(); plt.show()
```
