# Пара 6+. Wikidata SPARQL: Бунин и связанные сущности

Бесплатно, без ключа. Уважай лимиты: один спокойный запрос.


```python
import requests, pandas as pd

endpoint = "https://query.wikidata.org/sparql"
# Q233005 = Ivan Bunin
query = '''
SELECT ?item ?itemLabel ?typeLabel WHERE {
  VALUES ?bunin { wd:Q233005 }
  ?item ?p ?bunin .
  ?item wdt:P31 ?type .
  SERVICE wikibase:label { bd:serviceParam wikibase:language "ru,en". }
}
LIMIT 30
'''

r = requests.get(endpoint, params={"query": query, "format": "json"},
                 headers={"User-Agent": "SamGU-DH-Course/1.0 (teaching)"}, timeout=60)
r.raise_for_status()
bindings = r.json()["results"]["bindings"]
rows = [{
    "item": b["item"]["value"],
    "label": b.get("itemLabel", {}).get("value"),
    "type": b.get("typeLabel", {}).get("value"),
} for b in bindings]
df = pd.DataFrame(rows)
print(len(df), "rows")
df.head(20)
```

```python
# Простая визуализация: топ типов связанных сущностей
if len(df):
    top = df["type"].value_counts().head(10)
    ax = top.plot(kind="barh", figsize=(8,4), title="Типы сущностей, связанных с Буниным (Wikidata sample)")
    ax.invert_yaxis()
    import matplotlib.pyplot as plt
    plt.tight_layout(); plt.show()
else:
    print("Пусто — проверьте сеть / лимит SPARQL")
```
